```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load the Student Dropout dataset
df = pd.read_csv("clean.csv", sep=";")

# Check the dataset
print("Shape:", df.shape)

# Select numerical student-performance variables
cols = [
    "Curricular units 1st sem (approved)",
    "Curricular units 1st sem (grade)",
    "Curricular units 2nd sem (approved)",
    "Curricular units 2nd sem (grade)"
]

X = df[cols]

# Display the variables
print("Variables used:")
print(cols)

# Check the correlation between the variables
print("CORRELATION MATRIX:")
print(X.corr().round(2))

# Standardize the variables
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Check the shape after standardization
print("Scaled data shape:")
print(X_scaled.shape)

# Apply PCA
pca = PCA(n_components=2)
PC = pca.fit_transform(X_scaled)

# Show the variance explained by PC1 and PC2
print("EXPLAINED VARIANCE:")
print(pca.explained_variance_ratio_.round(3))

# Create a table containing PC1 and PC2
pca_df = pd.DataFrame(
    PC,
    columns=["PC1", "PC2"]
)

# Display the first 5 PCA scores
print("FIRST 5 PCA SCORES:")
print(pca_df.head())

# Add the student outcome after PCA
pca_df["Target"] = df["Target"]

# Compare PCA scores by student outcome
print("AVERAGE PCA SCORES BY STUDENT OUTCOME:")
print(
    pca_df.groupby("Target")[["PC1", "PC2"]]
    .mean()
    .round(2)
)
```
